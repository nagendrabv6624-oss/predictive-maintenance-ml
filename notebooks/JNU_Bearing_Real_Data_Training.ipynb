```python
# Step 3 — Display measured-data evaluation results
import json
from pathlib import Path
p=Path('outputs/jnu/metrics.json')
assert p.exists(), 'Training did not create metrics.json; inspect the preceding error.'
m=json.loads(p.read_text())
print('Dataset:',m['dataset'])
print('Split:',m['split_policy'])
for speed, models in m['results_by_held_out_speed'].items():
    print(f'\\nHeld-out speed: {speed} RPM')
    for name, result in models.items():
        print(f"{name}: accuracy={result['accuracy']:.3f}, macro-F1={result['macro_f1']:.3f}")
```
